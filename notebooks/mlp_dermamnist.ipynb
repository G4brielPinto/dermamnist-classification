# Multilayer Perceptron (MLP): Introduction

A **Multilayer Perceptron (MLP)** is a class of artificial neural networks composed of multiple layers of interconnected neurons. It is one of the most fundamental and widely used models in supervised learning, capable of solving both regression and classification tasks.

Unlike the **Simple Perceptron**, which can only model linearly separable data, an MLP introduces one or more **hidden layers** and applies non-linear **activation functions**, allowing the network to learn **complex, non-linear patterns** in data.

<br>

## General Structure and Operation

An MLP typically consists of:

- An **Input Layer**, which receives the raw input features.
- One or more **Hidden Layers**, where intermediate transformations are applied.
- An **Output Layer**, which produces the final prediction.

Each neuron in a layer is **fully connected** to all neurons in the subsequent layer. The neurons compute a **weighted sum** of their inputs, apply a **non-linear activation function** (such as the Sigmoid or ReLU), and pass the result forward to the next layer.

<br>

## Learning Process: Forward and Backward Propagation

The training of an MLP involves two key computational steps:

1. **Forward Propagation**:
   - Inputs are propagated through the network, layer by layer, producing an output prediction.
   - Each layer computes a transformation of the inputs using weights, biases, and activation functions.

2. **Backward Propagation (Backpropagation)**:
   - The error between the predicted output and the true label is propagated backward through the network.
   - Using the **chain rule** from calculus, gradients of the loss function are computed with respect to each weight.
   - These gradients are then used to **update the weights** via optimization algorithms such as **Stochastic Gradient Descent (SGD)** or **Adam**.

This **bidirectional flow**—forward for prediction and backward for learning—enables the MLP to iteratively improve its performance by minimizing a loss function (e.g., cross-entropy or mean squared error).

<br>

## Capabilities and Applications

Thanks to their structure and learning capacity, MLPs are capable of:

- Approximating any continuous function (Universal Approximation Theorem)
- Learning complex decision boundaries in high-dimensional feature spaces
- Handling both regression and classification tasks

**Common applications include**:
- Image recognition (when paired with flattening or feature extraction)
- Medical diagnosis (e.g., disease prediction from clinical parameters)
- Financial forecasting
- Text classification and natural language processing (NLP)
- Sensor data analysis and IoT systems

MLPs also serve as the conceptual foundation for more advanced architectures such as **Convolutional Neural Networks (CNNs)** and **Recurrent Neural Networks (RNNs)**.

## Forward Propagation: Matrix Representation and Computation

Let’s consider a specific case with:

- **3 input neurons**
- **2 hidden neurons**
- **1 output neuron**

To incorporate the **bias**, it is treated as an additional input with a constant value of 1. This means that each layer's weight matrix includes one extra column for the bias.

### Step 1️⃣ — From Input Layer to Hidden Layer

We define:

- Input vector (with bias):
  $$
  X =
  \begin{bmatrix}
  x_1 \\
  x_2 \\
  x_3 \\
  1
  \end{bmatrix}
  \quad (4 \times 1)
  $$

- Weight matrix for input to hidden layer:
  $$
  W^{[1]} =
  \begin{bmatrix}
  w_{11} & w_{12} & w_{13} & b_1 \\
  w_{21} & w_{22} & w_{23} & b_2
  \end{bmatrix}
  \quad (2 \times 4)
  $$

- Weighted sum for hidden layer:
  $$
  Z^{[1]} = W^{[1]} \cdot X
  $$

- Activation function (e.g., sigmoid) applied element-wise:
  $$
  H = a^{[1]} = \sigma(Z^{[1]})
  $$

So we obtain the hidden layer output vector $H$:

$$
H =
\begin{bmatrix}
h_1 \\
h_2
\end{bmatrix}
\quad (2 \times 1)
$$

We then **append a bias** neuron with value 1 to this vector:

$$
H_{\text{aug}} =
\begin{bmatrix}
h_1 \\
h_2 \\
1
\end{bmatrix}
\quad (3 \times 1)
$$

<br>

### Step 2️⃣ — From Hidden Layer to Output Layer

- Weight matrix from hidden to output layer:

$$
W^{[2]} =
\begin{bmatrix}
w_{31} & w_{32} & b_3
\end{bmatrix}
\quad (1 \times 3)
$$

- Weighted sum for output:

$$
Z^{[2]} = W^{[2]} \cdot H_{\text{aug}}
$$

- Apply activation function for final output:

$$
Y = a^{[2]} = \sigma(Z^{[2]})
$$

<br>

## Summary: General Formula for Each Layer

For a generic layer $l$ with input vector $a^{[l-1]}$ (with bias added), the forward pass is computed as:

$$
Z^{[l]} = W^{[l]} \cdot a^{[l-1]}
$$

$$
a^{[l]} = \sigma(Z^{[l]})
$$

This process is repeated for each layer until the output is produced.

<br>

The **forward propagation** enables the network to transform the input step-by-step through linear combinations and non-linearities, building a hierarchical understanding of the input features. The final output is then used during training to compute the loss and initiate backpropagation.

## Backward Propagation: Step-by-Step Theory and Computation

Once the **forward propagation** is complete, the next step in training a neural network is to propagate the **error backward** through the layers. This process is called **backpropagation**, and it enables the network to **learn** by adjusting the weights to minimize the **loss function**.

<br>

## Step-by-Step Guide to Backpropagation

Let’s build the theory of backpropagation in a modular and structured way, following the same design you used in your `forwardPropagation`.

<br>

### Overview of Backpropagation

The main goal of backpropagation is to **update the weights** of the network based on the **difference between predicted and actual outputs**, using the **gradient descent algorithm**.

We use the **chain rule** from calculus to compute how the loss changes with respect to each weight, and update accordingly.

<br>

## Notation and Structure

To keep things aligned with your implementation, we’ll use the following notations:

$$
L \text{: Total number of layers (including input, hidden, and output).}
$$

$$
a^{[l]} \text{: Activation vector of layer } l.
$$

$$
z^{[l]} \text{: Weighted input sum for layer } l, \text{ computed as } z^{[l]} = W^{[l]} \cdot a^{[l-1]}.
$$

$$
W^{[l]} \text{: Weight matrix connecting layer } l-1 \text{ to layer } l.
$$

$$
\delta^{[l]} \text{: Error at layer } l, \text{ used to compute gradients.}
$$

$$
\nabla W^{[l]} \text{: Gradient of the cost function with respect to weights } W^{[l]}.
$$

<br>

## Step 1 — Error at Output Layer

Assuming we use the **Mean Squared Error (MSE)** loss function:

$$
C = \frac{1}{2} \|a^{[L]} - y\|^2
$$

The error at the output layer \( L \) is:

$$
\delta^{[L]} = (a^{[L]} - y) \circ \sigma'(z^{[L]})
$$

Where:

$$
\circ \text{: denotes element-wise multiplication.}
$$

$$
\sigma' \text{: is the derivative of the activation function (e.g., sigmoid).}
$$

$$
y \text{: is the target (true) output.}
$$

This gives us the initial error to start the backward pass.

<br>

## Step 2 — Error at Hidden Layers

For each hidden layer \( l = L-1, L-2, ..., 2 \), the error is propagated backwards as:

$$
\delta^{[l]} = \left((W^{[l+1]})^T \cdot \delta^{[l+1]}\right) \circ \sigma'(z^{[l]})
$$

Important considerations:

$$
\text{- The bias weight added during forward pass must be excluded when multiplying with } W^T.
$$

$$
\text{- The derivative } \sigma'(z^{[l]}) \text{ must be computed using the pre-activation values } z^{[l]}, \text{ so make sure these are saved during forward propagation.}
$$

<br>

## Step 3 — Compute Weight Gradients

Once we have the error \( \delta^{[l]} \), we compute the gradient of the cost with respect to the weights:

$$
\nabla_W^{[l]} = \delta^{[l]} \cdot (a^{[l-1]}_{\text{with bias}})^T
$$

Which means:

$$
\text{- Multiply the error vector } \delta^{[l]} \text{ by the transposed activation vector from the previous layer (which includes the bias term).}
$$

<br>

## Step 4 — Update the Weights

Using the gradients, we update the weights with gradient descent:

$$
W^{[l]} := W^{[l]} - \eta \cdot \nabla_W^{[l]}
$$

Where:

$$
\eta \text{: is the learning rate, a small constant that controls the step size.}
$$

<br>

## Summary: Backpropagation Steps

Here is a compact checklist of the full backpropagation pipeline:

$$
1. \ \text{Input: Activations and weighted sums from forward pass + true output } y
$$

$$
2. \ \text{Output layer error: } \delta^{[L]} = (a^{[L]} - y) \circ \sigma'(z^{[L]})
$$

$$
3. \ \text{Hidden layer errors: } \delta^{[l]} = \left((W^{[l+1]})^T \cdot \delta^{[l+1]}\right) \circ \sigma'(z^{[l]})
$$

$$
4. \ \text{Weight gradients: } \nabla_W^{[l]} = \delta^{[l]} \cdot (a^{[l-1]}_{\text{with bias}})^T
$$

$$
5. \ \text{Weight updates: } W^{[l]} := W^{[l]} - \eta \cdot \nabla_W^{[l]}
$$

<br>

The **backpropagation** step is what enables learning in neural networks. It adjusts the model's internal parameters (weights) to reduce error and improve predictions over time.

```python
# Defining the Sigmoid activation function
def activationFunction(z):
    """
    Sigmoid activation function.

    Parameters:
        z (torch.Tensor): Input tensor

    Returns:
        torch.Tensor: Sigmoid activation output
    """
    activation = 1 / (1 + torch.exp(-z))
    return activation

# Derivative of the Sigmoid Function
def activationFunctionDerivative(z):
    """
    Derivative of the sigmoid activation function.

    Parameters:
        z (torch.Tensor): Input tensor

    Returns:
        torch.Tensor: Derivative of sigmoid at z
    """
    sigmoid = 1 / (1 + torch.exp(-z))
    return sigmoid * (1 - sigmoid)

# Function to compute the error
def error_function(activation, z, real_values, activation_derivative):
    """
    Compute the error for backpropagation.

    Parameters:
        activation (torch.Tensor): Activation output
        z (torch.Tensor): Pre-activation values
        real_values (torch.Tensor): Target values
        activation_derivative (function): Derivative of the activation function

    Returns:
        torch.Tensor: Computed error
    """
    error = (activation - real_values) * activation_derivative(z)
    return error

# First, let's define the new loss functions
def cross_entropy_loss(predictions, targets):
    """
    Compute the cross entropy loss.

    Parameters:
        predictions (torch.Tensor): Raw model output (logits)
        targets (torch.Tensor): Target values (one-hot or class indices)

    Returns:
        torch.Tensor: Cross entropy loss value
    """
    # Convert targets to class indices if they're one-hot encoded
    if targets.shape == predictions.shape:
        targets = torch.argmax(targets, dim=0)

    # Apply log softmax to predictions
    log_softmax = torch.log_softmax(predictions, dim=0)

    # Cross entropy loss is -log(p) for the correct class
    loss = -log_softmax[targets]

    return loss.mean()

def multi_margin_loss(predictions, targets, margin=1.0):
    """
    Compute the multi-class margin loss.

    Parameters:
        predictions (torch.Tensor): Raw model output (logits)
        targets (torch.Tensor): Target values (one-hot or class indices)
        margin (float): Margin parameter

    Returns:
        torch.Tensor: Multi-margin loss value
    """
    # Convert targets to class indices if they're one-hot encoded
    if targets.shape == predictions.shape:
        targets = torch.argmax(targets, dim=0)

    # Get the score for the correct class
    correct_class_score = predictions[targets]

    # Compute margins against all other classes
    margins = torch.clamp(predictions - correct_class_score + margin, min=0)

    # Zero out the margin for the correct class
    margins[targets] = 0

    return margins.sum() / max(1, (margins > 0).sum())
```

```python
class NeuralNetwork:
    """
    Multilayer Perceptron (MLP) implementation with custom forward and backward propagation.

    This class implements a fully connected neural network with configurable architecture and
    training procedures optimized for PyTorch.
    """

    def __init__(self, input_neurons, hidden_neurons, output_neurons, loss_function='mse', optimizer_type='sgd', device=None):
        """
        Initialize the neural network with the specified architecture.

        Parameters:
            input_neurons (int): Number of input features
            hidden_neurons (list): List containing the number of neurons in each hidden layer
            output_neurons (int): Number of output neurons (classes)
            loss_function (str): Loss function to use ('mse', 'cross_entropy', or 'multi_margin')
            optimizer_type (str): Optimizer to use ('sgd' or 'adam')
            device (torch.device): Device to use for computation (CPU or CUDA)
        """
        self.input_neurons = input_neurons
        self.hidden_neurons = hidden_neurons
        self.output_neurons = output_neurons

        # Set loss function
        self.loss_function = loss_function

        # Set optimizer type (can be None)
        self.optimizer_type = optimizer_type

        # Flag to determine if we should apply activation in the output layer
        self.apply_output_activation = (loss_function == 'mse')

        # Set computation device (GPU if available, CPU otherwise)
        self.device = device if device is not None else torch.device("cuda" if torch.cuda.is_available() else "cpu")
        print(f"Using device: {self.device}")
        print(f"Using loss function: {self.loss_function}")
        print(f"Using optimizer: {self.optimizer_type}")

        # Initialize weight matrices for each layer with uniform random values
        self.weights = []

        # Input Layer to 1st Hidden Layer
        # Rows: neurons in the first hidden layer
        # Columns: neurons in the input layer plus an additional column for the bias term
        self.weights.append(
            torch.empty(hidden_neurons[0], input_neurons + 1, device=self.device).uniform_(-0.5, 0.5)
        )

        # Weights between consecutive hidden layers
        for idx_hid_layer in range(len(hidden_neurons) - 1):
            # Rows: neurons in the next hidden layer
            # Columns: neurons in the previous hidden layer plus bias
            self.weights.append(
                torch.empty(hidden_neurons[idx_hid_layer + 1], hidden_neurons[idx_hid_layer] + 1, device=self.device).uniform_(-0.5, 0.5)
            )

        # Last Hidden Layer to Output Layer
        # Rows: neurons in the output layer
        # Columns: neurons in the last hidden layer plus bias
        self.weights.append(
            torch.empty(output_neurons, hidden_neurons[-1] + 1, device=self.device).uniform_(-0.5, 0.5)
        )

        # Initialize the optimizer if an optimizer type is specified
        self.optimizer = None
        if self.optimizer_type:
            if self.optimizer_type == 'sgd':
                self.optimizer = torch.optim.SGD([w.requires_grad_(True) for w in self.weights], lr=0.01, momentum=0.9)
            elif self.optimizer_type == 'adam':
                self.optimizer = torch.optim.Adam([w.requires_grad_(True) for w in self.weights], lr=0.001)
            else:
                raise ValueError(f"Unknown optimizer type: {self.optimizer_type}")

    def compute_gradients(self, activations, zs, target):
        """
        Compute gradients for the weights based on the error and activations.

        Parameters:
            activations (list): List of activation values from forward propagation
            zs (list): List of weighted sum values from forward propagation
            target (torch.Tensor): Target output values

        Returns:
            list: List of gradient tensors for each weight matrix
        """
        # List to save each layer error
        delta = []

        # List to save each gradient of weight
        gradient_weights = []

        # 1. Compute the error on the output layer based on loss function
        if self.loss_function == 'mse':
            # Standard MSE error calculation with sigmoid activation
            error_out_layer = error_function(activations[-1], zs[-1], target, activationFunctionDerivative)
        else:
            # For cross_entropy and multi_margin, compute gradient directly
            if target.shape == activations[-1].shape:
                target_class = torch.argmax(target, dim=0)
            else:
                target_class = target.item()

            # Initialize gradient as zeros
            error_out_layer = torch.zeros_like(activations[-1])

            if self.loss_function == 'cross_entropy':
                # Softmax gradient for cross entropy
                softmax_output = torch.softmax(activations[-1], dim=0)
                softmax_output[target_class] -= 1.0
                error_out_layer = softmax_output
            else:  # multi_margin
                # Compute gradient for multi-margin loss
                margin = 1.0
                for i in range(self.output_neurons):
                    if i != target_class and (activations[-1][i] - activations[-1][target_class] + margin > 0):
                        error_out_layer[i] += 1.0
                        error_out_layer[target_class] -= 1.0

            # Normalize by the number of non-zero terms
            num_active = max(1, torch.sum(error_out_layer != 0) / 2)
            error_out_layer /= num_active

        delta.append(error_out_layer)

        # 2. Compute the error on each hidden layer (backwards)
        for hidden_layer_idx in range(len(self.weights) - 2, -1, -1):
            # Remove the bias weight for correct error propagation
            weight_no_bias = self.weights[hidden_layer_idx + 1][:, :-1]
            weighted_error = torch.matmul(weight_no_bias.t(), delta[-1])

            # Apply Hadamard product with activation derivative
            error_hidden_layer_idx = weighted_error * activationFunctionDerivative(zs[hidden_layer_idx])

            # Add the error to delta list for next iteration
            delta.append(error_hidden_layer_idx)

        # 3. Compute the gradients of weights
        # Reverse delta order to align with layer order
        delta.reverse()
        for hidden_layer_idx in range(len(self.weights)):
            # Augment activation with bias term
            activation_with_bias = torch.cat([activations[hidden_layer_idx], torch.ones(1, 1, device=self.device)], dim=0)

            # Compute gradient as outer product of error and activation
            gradient_weight = torch.matmul(delta[hidden_layer_idx], activation_with_bias.t())

            # Store gradient
            gradient_weights.append(gradient_weight)

        return gradient_weights

    def forwardPropagation(self, data_input):
        """
        Perform forward propagation through the network.

        Parameters:
            data_input (torch.Tensor): Input data tensor of shape [input_neurons, 1]

        Returns:
            tuple: (activations, weighted_sums)
                - activations: List of activation values for each layer
                - weighted_sums: List of weighted sum values before activation
        """
        # Verify input data shape
        if (data_input.shape[0] != self.input_neurons) or (data_input.shape[1] != 1):
            print(f"ERROR: The shape of the input data is incorrect. Expected {(self.input_neurons, 1)}, got {data_input.shape}")
            return None, None

        # List to store activation vectors of each layer (including input)
        activations = []
        zs = []  # Store weighted sums for backpropagation

        # Initial activation is simply the input data
        activation = data_input
        activations.append(activation)

        # Forward pass through each layer
        for i, weight_matrix in enumerate(self.weights):
            # Augment activation vector by adding the bias term (value 1)
            activation_with_bias = torch.cat([activation, torch.ones(1, 1, device=self.device)], dim=0)

            # Linear combination: Z = W · A (bias included in W and 1 in A)
            Z = torch.matmul(weight_matrix, activation_with_bias)
            zs.append(Z)

            # Apply the activation function (sigmoid) except for the output layer with non-MSE loss
            if i < len(self.weights) - 1 or self.apply_output_activation:
                activation = activationFunction(Z)
            else:
                # For cross_entropy and multi_margin, use the raw logits
                activation = Z

            # Store the activation of the current layer
            activations.append(activation)

        return activations, zs

    def compute_loss(self, output, target):
        """
        Compute the loss based on the selected loss function.

        Parameters:
            output (torch.Tensor): Model output
            target (torch.Tensor): Target values

        Returns:
            torch.Tensor: Loss value
        """
        if self.loss_function == 'mse':
            return torch.mean((output - target) ** 2)
        elif self.loss_function == 'cross_entropy':
            return cross_entropy_loss(output, target)
        elif self.loss_function == 'multi_margin':
            return multi_margin_loss(output, target)
        else:
            raise ValueError(f"Unknown loss function: {self.loss_function}")

    def backwardPropagation(self, activations, zs, target, learning_rate=0.01):
        """
        Perform backward propagation to update weights based on computed error.

        Parameters:
            activations (list): List of activation values from forward propagation
            zs (list): List of weighted sum values from forward propagation
            target (torch.Tensor): Target output values
            learning_rate (float): Learning rate for weight updates
        """
        # Compute gradients
        gradients = self.compute_gradients(activations, zs, target)

        # Update weights using the selected optimizer if available
        if self.optimizer_type and self.optimizer:
            # Clear previous gradients
            self.optimizer.zero_grad()

            # Manually update gradients (since we computed them ourselves)
            for i, weight in enumerate(self.weights):
                if weight.grad is None:
                    weight.grad = gradients[i].clone()
                else:
                    weight.grad.copy_(gradients[i])

            # Step the optimizer
            self.optimizer.step()
        else:
            # Manual weight update without optimizer
            for hidden_layer_idx in range(len(self.weights)):
                self.weights[hidden_layer_idx] -= learning_rate * gradients[hidden_layer_idx]

    def train(self, training_data, epochs=1000, learning_rate=0.01, print_every=100, batch_size=None, validation_data=None):
        """
        Train the neural network using mini-batch gradient descent.

        Parameters:
            training_data (DataLoader): DataLoader containing training data batches
            epochs (int): Number of training epochs
            learning_rate (float): Learning rate for weight updates
            print_every (int): Frequency of progress reporting
            batch_size (int): Mini-batch size (None for full-batch)
            validation_data (DataLoader): Optional validation dataset

        Returns:
            tuple: (training_losses, validation_accuracies)
        """
        training_losses = []  # Store training loss for visualization
        validation_accuracies = []  # Track validation accuracy

        # Update learning rate for the optimizer if it exists
        if self.optimizer_type and self.optimizer:
            if self.optimizer_type == 'sgd':
                self.optimizer = torch.optim.SGD([w.requires_grad_(True) for w in self.weights], lr=learning_rate, momentum=0.9)
            elif self.optimizer_type == 'adam':
                self.optimizer = torch.optim.Adam([w.requires_grad_(True) for w in self.weights], lr=learning_rate)

        for epoch in range(epochs):
            total_loss = 0
            total_samples = 0

            # Process each batch from the DataLoader
            for batch_inputs, batch_targets in training_data:
                batch_loss = 0

                # Process each sample in the batch
                for i in range(len(batch_inputs)):
                    # Extract single input and target
                    input_vector = batch_inputs[i].view(-1, 1).float().to(self.device)

                    # Convert target to one-hot encoding if needed (for MSE loss)
                    if batch_targets[i].numel() == 1:
                        if self.loss_function == 'mse':
                            target_vector = torch.zeros(self.output_neurons, 1, device=self.device)
                            target_vector[batch_targets[i].item()] = 1.0
                        else:
                            # For cross_entropy and multi_margin, keep as class index
                            target_vector = batch_targets[i].to(self.device)
                    else:
                        target_vector = batch_targets[i].view(-1, 1).float().to(self.device)

                    # 1. Forward Propagation
                    activations, zs = self.forwardPropagation(input_vector)

                    # 2. Backward Propagation
                    self.backwardPropagation(activations, zs, target_vector, learning_rate)

                    # 3. Loss Calculation
                    loss = self.compute_loss(activations[-1], target_vector)
                    batch_loss += loss.item()

                # Update total loss
                total_loss += batch_loss
                total_samples += len(batch_inputs)

            # Calculate average training loss for this epoch
            avg_loss = total_loss / total_samples
            training_losses.append(avg_loss)

            # Validate if validation data is provided
            if validation_data:
                with torch.no_grad():  # Disable gradient tracking during validation
                    correct = 0
                    total = 0

                    for val_inputs, val_targets in validation_data:
                        # Process each validation sample
                        for i in range(len(val_inputs)):
                            # Extract input and target
                            val_input = val_inputs[i].view(-1, 1).float().to(self.device)
                            val_target = val_targets[i].to(self.device)

                            # Forward pass
                            output = self.forwardPropagation(val_input)[0][-1]

                            # Get predicted class and true class
                            predicted_class = torch.argmax(output).item()
                            true_class = val_target.item() if val_target.numel() == 1 else torch.argmax(val_target).item()

                            # Count correct predictions
                            if predicted_class == true_class:
                                correct += 1
                            total += 1

                    val_accuracy = correct / total
                    validation_accuracies.append(val_accuracy)
            else:
                val_accuracy = None

            # Report progress periodically
            if (epoch + 1) % print_every == 0 or epoch == 0:
                print(
                    f"{Fore.CYAN}Epoch {epoch + 1:4d}/{epochs}{Style.RESET_ALL} | "
                    f"{Fore.YELLOW}Train Loss: {avg_loss:.6f}{Style.RESET_ALL} " +
                    (f"| {Fore.GREEN}Val Accuracy: {val_accuracy*100:.2f}%{Style.RESET_ALL}" if val_accuracy is not None else "")
                )

        # Visualization of training progress
        plt.figure(figsize=(10, 5))

        # Plot training loss
        ax1 = plt.subplot(1, 1 if not validation_accuracies else 2, 1)
        ax1.plot(training_losses, label='Training Loss', color='blue')
        ax1.set_xlabel("Epoch")
        ax1.set_ylabel("Loss")
        ax1.set_title("Training Loss")
        ax1.legend()
        ax1.grid(True)

        # Plot validation accuracy if available
        if validation_accuracies:
            ax2 = plt.subplot(1, 2, 2)
            ax2.plot(validation_accuracies, label='Validation Accuracy', color='green')
            ax2.set_xlabel("Epoch")
            ax2.set_ylabel("Accuracy")
            ax2.set_title("Validation Accuracy")
            ax2.legend()
            ax2.grid(True)

        plt.tight_layout()
        plt.show()

        return training_losses, validation_accuracies
```

```python
# preprocessing
data_transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                     std=[0.229, 0.224, 0.225])
])

# Load the official training, validation, and test splits
train_dataset = DataClass(split='train', transform=data_transform, download=download)
val_dataset = DataClass(split='val', transform=data_transform, download=download)
test_dataset = DataClass(split='test', transform=data_transform, download=download)


class_distribution = [228, 359, 769, 80, 779, 4693, 99]
total_samples = sum(class_distribution)
class_weights = [total_samples / (len(class_distribution) * count) for count in class_distribution]
class_weights = torch.FloatTensor(class_weights)
print("Class Weights:", class_weights)

# Create sampler based on class distribution
class_indices = [[] for _ in range(len(class_distribution))]
temp_loader = DataLoader(train_dataset, batch_size=128, shuffle=False)
start_idx = 0
for batch_data, batch_labels in temp_loader:
    for i, label in enumerate(batch_labels):
        class_idx = label.item()
        class_indices[class_idx].append(start_idx + i)
    start_idx += len(batch_labels)

sample_weights = torch.zeros(len(train_dataset))
for class_idx, indices in enumerate(class_indices):
    for idx in indices:
        sample_weights[idx] = class_weights[class_idx]

sampler = WeightedRandomSampler(sample_weights, len(sample_weights), replacement=True)
train_loader = DataLoader(dataset=train_dataset, batch_size=BATCH_SIZE, sampler=sampler)
train_loader_at_eval = DataLoader(dataset=train_dataset, batch_size=2*BATCH_SIZE, shuffle=False)
val_loader = DataLoader(dataset=val_dataset, batch_size=2*BATCH_SIZE, shuffle=False)
test_loader = DataLoader(dataset=test_dataset, batch_size=2*BATCH_SIZE, shuffle=False)
```

```python
def plot_dataset_distribution(train_dataset, test_dataset, class_names=None):
    """
    Plot the class distribution in training and test datasets.

    Parameters:
        train_dataset: Training dataset object
        test_dataset: Test dataset object
        class_names (list): Optional list of class names
    """
    def extract_label(item):
        # Extract the label from various possible formats
        if isinstance(item, tuple):
            label = item[1]  # (image, label) tuple
        else:
            label = item

        if isinstance(label, torch.Tensor):
            if label.ndim > 0 and label.numel() > 1:
                return torch.argmax(label).item()
            else:
                return label.item()
        elif hasattr(label, 'numpy') and label.ndim > 0:
            return int(label.argmax())  # NumPy array
        return int(label)

    # Check if input is a DataLoader and extract the dataset if needed
    train_dataset = train_dataset.dataset if hasattr(train_dataset, 'dataset') else train_dataset
    test_dataset = test_dataset.dataset if hasattr(test_dataset, 'dataset') else test_dataset

    # Extract labels as integers
    train_labels = [extract_label(train_dataset[i][1]) for i in range(len(train_dataset))]
    test_labels = [extract_label(test_dataset[i][1]) for i in range(len(test_dataset))]

    # Count frequency of each class
    train_counts = Counter(train_labels)
    test_counts = Counter(test_labels)

    # Get all unique classes
    all_classes = sorted(set(train_counts.keys()) | set(test_counts.keys()))

    # Create class names if not provided
    if class_names is None:
        class_names = [str(cls) for cls in all_classes]

    train_values = [train_counts.get(cls, 0) for cls in all_classes]
    test_values = [test_counts.get(cls, 0) for cls in all_classes]

    # Plot class distribution
    x = range(len(all_classes))
    width = 0.35

    plt.figure(figsize=(12, 6))
    plt.bar(x, train_values, width, label='Training', color='royalblue')
    plt.bar([i + width for i in x], test_values, width, label='Test', color='salmon')
    plt.xlabel('Class')
    plt.ylabel('Number of samples')
    plt.title('Class Distribution in Dataset')
    plt.xticks([i + width/2 for i in x], class_names, rotation=45, ha='right')
    plt.legend()
    plt.tight_layout()
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.show()

def evaluate_model(mlp, test_loader, class_names=None):
    """
    Evaluate the MLP model on test data and display metrics and visualizations.

    Parameters:
        mlp: Trained MLP model
        test_loader: DataLoader containing test data
        class_names (list): Optional list of class names
    """
    # Collect predictions
    y_true = []
    y_pred = []
    pred_probs = []

    with torch.no_grad():
        for inputs, targets in test_loader:
            # Process each test sample
            for i in range(len(inputs)):
                # Extract input and target
                input_tensor = inputs[i].view(-1, 1).float().to(mlp.device)
                target = targets[i].to(mlp.device)

                # Forward pass
                output = mlp.forwardPropagation(input_tensor)[0][-1]

                # Get predicted class and true class
                predicted = torch.argmax(output).item()
                true_label = target.item() if target.numel() == 1 else torch.argmax(target).item()

                # Store results
                y_pred.append(predicted)
                y_true.append(true_label)
                pred_probs.append(output.cpu().numpy())  # Store probabilities for ROC curve

    # Calculate evaluation metrics
    accuracy = accuracy_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred, average='weighted', zero_division=0)
    recall = recall_score(y_true, y_pred, average='weighted', zero_division=0)
    f1 = f1_score(y_true, y_pred, average='weighted', zero_division=0)

    # Display metrics
    print("\n===== Model Evaluation =====")
    print(f"Accuracy:  {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall:    {recall:.4f}")
    print(f"F1-Score:  {f1:.4f}")

    # Plot confusion matrix
    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(10, 8))
    sns.heatmap(
        cm,
        annot=True,
        fmt='d',
        cmap="Blues",
        xticklabels=class_names if class_names else None,
        yticklabels=class_names if class_names else None
    )
    plt.xlabel("Predicted")
    plt.ylabel("True")
    plt.title("Confusion Matrix")
    plt.tight_layout()
    plt.show()

    return {
        'accuracy': accuracy,
        'precision': precision,
        'recall': recall,
        'f1': f1,
        'confusion_matrix': cm,
        'y_true': y_true,
        'y_pred': y_pred,
        'pred_probs': pred_probs
    }

def show_sample_predictions(mlp, test_dataset, n_samples=8, class_names=None):
    """
    Display sample images with their predictions.

    Parameters:
        mlp: Trained MLP model
        test_dataset: Test dataset object
        n_samples (int): Number of samples to show
        class_names (list): Optional list of class names
    """
    # Check if test_dataset is a DataLoader and extract the dataset if needed
    if hasattr(test_dataset, 'dataset'):
        test_dataset = test_dataset.dataset

    # Randomly select samples
    idxs = np.random.choice(len(test_dataset), size=n_samples, replace=False)

    plt.figure(figsize=(15, 8))

    for i, idx in enumerate(idxs):
        # Get sample
        img, label = test_dataset[idx]

        # Prepare for prediction
        img_input = img.view(-1, 1).float().to(mlp.device)

        # Predict
        with torch.no_grad():
            output = mlp.forwardPropagation(img_input)[0][-1]
            predicted = torch.argmax(output).item()
            confidence = output[predicted].item()

        # Display
        plt.subplot(2, n_samples//2, i+1)

        # Show image (account for different channel dimensions)
        if img.shape[0] == 1:  # Grayscale
            plt.imshow(img.squeeze().numpy(), cmap='gray')
        else:  # RGB
            plt.imshow(img.permute(1, 2, 0).numpy())

        # Create title with prediction info
        label_text = f"{class_names[label] if class_names else label}" if isinstance(label, int) else f"{class_names[label.item()] if class_names else label.item()}"
        pred_text = f"{class_names[predicted] if class_names else predicted}"
        title = f"Pred: {pred_text}\nTrue: {label_text}\nConf: {confidence:.2f}"

        # Color based on correctness
        true_label = label if isinstance(label, int) else label.item()
        color = 'green' if predicted == true_label else 'red'
        plt.title(title, color=color)
        plt.axis('off')

    plt.tight_layout()
    plt.show()
```

```python
def run_experiment(hidden_layers, loss_function, optimizer_type, learning_rate, epochs, batch_size):
    """Train one MLP configuration and evaluate it on validation data.

    Args:
        hidden_layers (list): Number of neurons in each hidden layer.
        loss_function (str): Loss function name: mse, cross_entropy, or multi_margin.
        optimizer_type (str): Optimizer name: sgd, adam, or None.
        learning_rate (float): Learning rate used during training.
        epochs (int): Maximum number of training epochs.
        batch_size (int): Training batch size.

    Returns:
        dict: Model, validation metrics, and training history.
    """
    print(f"\n{'=' * 50}")
    print("EXPERIMENT")
    print(f"Hidden layers: {hidden_layers}")
    print(f"Loss function: {loss_function}")
    print(f"Optimizer: {optimizer_type}")
    print(f"Learning rate: {learning_rate}")
    print(f"Epochs: {epochs}")
    print(f"Batch size: {batch_size}")
    print(f"{'=' * 50}\n")

    model = NeuralNetwork(
        info["n_channels"] * 28 * 28,
        hidden_layers,
        n_classes,
        loss_function=loss_function,
        optimizer_type=optimizer_type,
        device=torch.device("cuda" if torch.cuda.is_available() else "cpu"),
    )

    train_losses, val_accuracies = model.train(
        training_data=train_loader,
        epochs=epochs,
        learning_rate=learning_rate,
        print_every=5,
        batch_size=batch_size,
        validation_data=val_loader,
    )

    print("\nValidation evaluation:")
    validation_metrics = evaluate_model(
        model,
        val_loader,
        class_names=list(info["label"].values()),
    )

    return {
        "model": model,
        "validation_metrics": validation_metrics,
        "train_loss": train_losses,
        "val_accuracy": val_accuracies,
    }


def evaluate_final_model(model):
    """Evaluate a selected model once on the held-out test split."""
    return evaluate_model(model, test_loader, class_names=list(info["label"].values()))
```
